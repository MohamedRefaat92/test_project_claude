```python
# Example: verify CLI tools are available
import subprocess

tools = ['samtools', 'STAR', 'kallisto', 'salmon', 'fastqc', 'bcftools']
for tool in tools:
    result = subprocess.run(['which', tool], capture_output=True, text=True)
    status = result.stdout.strip() if result.returncode == 0 else 'NOT FOUND'
    print(f"  {tool:12s}: {status}")
```
